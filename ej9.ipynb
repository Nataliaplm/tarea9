```python
# Dataset

tareas_sprint = [
{"id": "T-01", "historia": "Login de usuarios", "puntos": 5, "estado": "Done",
"responsable": "Marta"},
{"id": "T-02", "historia": "Base de datos clientes", "puntos": 8, "estado": "Done",
"responsable": "Carlos"},
{"id": "T-03", "historia": "API de pagos", "puntos": 13, "estado": "In Progress",
"responsable": "Marta"},
{"id": "T-04", "historia": "Dashboard inicial", "puntos": 5, "estado": "done",
"responsable": "Lucia"},
{"id": "T-05", "historia": "Exportar reporte CSV", "puntos": 3, "estado": "To Do",
"responsable": "Marta"},
{"id": "T-06", "historia": "Pruebas de seguridad", "puntos": 5, "estado": "In Progress",
 "responsable": None},
{"id": "T-07", "historia": "Documentacion API", "puntos": 2, "estado": "Done",
"responsable": "Lucia"},
{"id": "T-08", "historia": "Optimizacion consultas", "puntos": 8, "estado": "To Do",
 "responsable": "Marta"}
]
```

```python
# Limpiar datos y detectar si existen tareas huérfanas sin responsable asignado (None).

nones=0
for tarea in tareas_sprint:
    tarea["estado"] = tarea["estado"].lower() 
    if tarea["responsable"] == None:
        nones+=1
valor = (nones/len(tareas_sprint))*100
print(valor, "% de las tareas no tienen dueño")  
```

```python
#Medir avance del equipo
puntos = 0
completados=0
for tarea in tareas_sprint:
    puntos += tarea["puntos"]
    if tarea["estado"] == "done":
        completados += tarea["puntos"]

print("puntos totales: ", puntos)
print("puntos completados: ", completados)
avance= round((completados/puntos*100),2) 
print(avance, "% de avance en la tarea")
```

```python
# Inicializamos los acumuladores
marta = 0
carlos = 0
lucia = 0
sin_responsable = 0

for tarea in tareas_sprint:
    if tarea["responsable"] == "Marta":
        marta += tarea["puntos"]
    elif tarea["responsable"] == "Carlos":
        carlos += tarea["puntos"]
    elif tarea["responsable"] == "Lucia":
        lucia += tarea["puntos"]
    elif tarea["responsable"] is None:
        sin_responsable += tarea["puntos"]

total = marta + carlos + lucia + sin_responsable

porcentaje_marta = marta / total * 100
porcentaje_carlos = carlos / total * 100
porcentaje_lucia = lucia / total * 100
print("Marta:", marta, "tareas,",  porcentaje_marta, "%")
print("Carlos:", carlos, "tareas,", porcentaje_carlos, "%")
print("Lucia:", lucia, "tareas,", porcentaje_lucia, "%")

if porcentaje_marta > 40:
    print("Marta concentra más del 40% del trabajo")
if porcentaje_carlos > 40:
    print("Carlos concentra más del 40% del trabajo")
if porcentaje_lucia > 40:
    print("Lucia concentra más del 40% del trabajo")
```
